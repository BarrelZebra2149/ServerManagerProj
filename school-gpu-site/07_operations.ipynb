## A. 웹 터미널 (🌐)
- 학생: **웹 터미널** 메뉴(자기 서버), 관리자: **서버 상세 → 웹 터미널**(어느 서버든 자기 리눅스 계정).
- 서버에는 **5분짜리 임시 SSH 인증서**로 접속합니다(비밀번호·키를 사이트에 저장하지 않음). 30분간 입력이 없으면 종료됩니다.
- **복사·붙여넣기**(MobaXterm 방식): 드래그로 선택하면 자동 복사, **오른쪽 클릭**으로 붙여넣기, `Ctrl+Shift+C` / `Ctrl+Shift+V`. `Ctrl+C` 는 그대로 실행 중단. **여러 줄**을 붙여넣으면 명령이 바로 실행될 수 있어 확인창이 뜹니다(설정에서 끌 수 있음).

## B. 파일 탐색기 (웹 터미널 왼쪽 패널)
| 기능 | 사용법 |
|---|---|
| 목록 | 폴더가 먼저, 이름순. 파일 위에 마우스를 올리면 수정 시각. **숨김 파일** 토글 |
| 이동 | 폴더 클릭, `↑` 상위, 위의 경로 칸에 직접 입력 |
| **cd 따라가기** | 터미널에서 `cd` 하면 목록도 그 폴더로 이동(서버에 셸 연동 필요, 아래 C) |
| 열기·수정·저장 | 텍스트 파일을 누르면 에디터가 열리고 **Ctrl+S** 로 저장. 1 MB 초과·이진·UTF-8 이 아닌 파일은 열리지 않음. 다른 곳에서 바뀐 파일은 **충돌 경고** 후 덮어쓰기 여부를 물음 |
| 내려받기 | 파일의 `⋯` 메뉴(또는 오른쪽 클릭) → 내려받기. 100 MB 이하 |
| 올리기 | **끌어다 놓기** 또는 `올리기` 버튼. 100 MB 이하, 같은 이름은 덮어쓰기 확인 |
| 새 폴더·이름 바꾸기·삭제 | 삭제는 파일과 **빈 폴더만**(찬 폴더는 터미널에서). 확인창 필수 |

권한은 **로그인한 리눅스 계정 그대로**입니다(읽을 수 없는 파일은 열리지 않음). 파일 변경은 감사 로그에 경로만 남고 내용은 남지 않습니다.
큰 폴더 전체는 학생 대시보드의 **작업 폴더 백업(tar.gz)** 이나 `scp`/`rsync` 를 쓰세요.

## C. cd 따라가기 — 셸 연동 설치
셸이 현재 폴더를 알려 주는 표준 신호(OSC 7)를 보내도록 `/etc/profile.d/ssm-osc7.sh` 한 파일을 설치합니다.
- **새로 가입·재가입하는 서버**: 가입 스크립트가 자동 설치.
- **이미 가입한 서버**: **서버 상세 → 셸 연동 설치** 버튼(관리자). 이미 열린 터미널은 다시 로그인해야 적용됩니다.
  - 에이전트가 **옛 버전**이면(에이전트 0.2.0 이전) "에이전트가 이 기능을 지원하지 않는 옛 버전" 메시지가 뜹니다. **재가입을 딱 한 번** 하면 이후로는 **서버 상세 → 에이전트 업데이트** 버튼(재가입 없이 에이전트가 스스로 새 버전을 설치하고 재시작)으로 갱신합니다.
- 신호가 안 오면 파일 패널에 "터미널 위치 신호가 아직 없습니다"가 보이고, 목록은 홈에서 시작해 클릭으로만 이동합니다.

## D. SSH 키로 `ssh myserver` (학생, 교내)
학생 대시보드의 **SSH 키 · 자동 로그인**에서 OS(Windows/macOS·Linux)에 맞는 스크립트를 내 PC 에서 한 번 실행하면 `~/.ssh/config` 에 `Host myserver` 가 생기고 키가 만들어집니다(비밀키는 사이트로 가지 않음). 만들어진 **공개키 한 줄**을 사이트에 붙여 넣어 등록하면 이후 `ssh myserver` 한 번으로 접속됩니다. **학교 밖**에서는 HTTPS 만 열려 있으므로 웹 터미널을 씁니다.

## F. GPU 가 멈췄을 때 — 강제 재시작
현상: 무거운 작업 중 GPU 가 응답하지 않고 `nvidia-smi` 도 멈춤. **`sudo shutdown now` 는 쓰지 않습니다**(서버를 직접 켜야 함).
**서버 현황**에서 빨간 "GPU 응답 없음" 카드 → **강제 재시작** → 서버 이름 직접 입력 + 사유 + 비밀번호·인증 코드.

| 단계 | 방법 |
|---|---|
| 1 | 에이전트가 `s`(디스크 동기화) → `u`(읽기 전용 재마운트) → `b`(재부팅) |
| 2 | 에이전트가 응답하지 않으면 관리 서버가 **재부팅 전용 SSH 키**로 접속(키는 재부팅 외엔 못 하도록 제한) |
| 3 | 둘 다 실패하면 "현장 전원 조작 필요" |

서버에서 직접(콘솔·SSH 가 될 때)은 `echo 1 | sudo tee /proc/sys/kernel/sysrq` 후 `s`, `u`, `b` 를 `/proc/sysrq-trigger` 에 순서대로 씁니다. **빈 서버에서 먼저 시험**해 두세요.
